## 2. The question: decision, action, cost of a wrong call

**Decision:** Which content items should the refresh team review first?
**Action:** A human reviewer opens the top-ranked pages and applies refresh, expansion, protection, pruning, or monitoring actions.
**Cost of a wrong call:** Wasted reviewer hours on pages that don't need action, plus missed declines on pages that do. With limited review capacity, ordering by expected impact is critical — a wrong top-K order means the team spends time on low-value pages while high-value declines go unaddressed.

```python
# Load starter CSV and show 2-3 real numbers
import pandas as pd, json
df = pd.read_csv('data/raw/content_refresh_anonymized.csv')
m = json.load(open('data/processed/feature_metadata.json'))
print('Rows in starter data:', len(df))
print('Rows marked declining (trend_direction == down):', int(df['trend_direction'].str.lower().eq('down').sum()))
print('Declining rate:', round(df['trend_direction'].str.lower().eq('down').mean(), 3))
print('Features available:', len(m['model_numeric_features']), 'numeric +', len(m['model_categorical_features']), 'categorical')
print('Sample: impressions_90d mean =', round(df['impressions_90d'].mean(), 0), ', sessions_90d mean =', round(df['sessions_90d'].mean(), 1))
```
